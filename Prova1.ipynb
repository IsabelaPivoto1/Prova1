```python
nome = input('Qual é o seu nome? ')

idade = int(input('Qual é a sua idade? '))

dinheiro = float(input('Quanto dinheiro você tem (R$)? '))


if idade < 16:
    print(f'{nome}, Entrada permitida apenas com acompanhante')
else:

    if dinheiro >= 30.0:
        print(f'{nome}, Ingresso comprado! Bom filme.')
    else:
        print(f'Desculpe {nome}, Saldo insuficiente para o ingresso.')


print('Foi bom te atender!!')
```
